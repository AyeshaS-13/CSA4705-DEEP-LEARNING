Q1.Vector Addition, Subtraction and Scalar Multiplication

```python
import numpy as np

# Initial robot position
position = np.array([2, 3, 4])

# Movement vector
movement = np.array([5, 2, 1])

# Vector addition
new_position = position + movement

# Vector subtraction
difference = position - movement

# Scalar multiplication
scaled_position = 2 * position

print("Initial Position:", position)
print("Movement Vector:", movement)
print("New Position:", new_position)
print("Vector Subtraction:", difference)
print("Scalar Multiplication:", scaled_position)
```
